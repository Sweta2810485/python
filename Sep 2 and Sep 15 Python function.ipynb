```python
x = 10  # Global variable

def update_x():
    global x  # Declare that we are using the global variable x
    x += 5    # Modify the global variable
   
update_x()
print(x)  # Output: 15
```

```python
x = 10  # Global variable

def update_x():
    globals()['x'] += 5  # Access and modify the global variable 

update_x()
print(x)  # Output: 15
```

```python
def count(lst):
  
    even = 0
    odd = 0
    
    for i in lst:
        if i%2 == 0:
            even += 1
        else:
            odd +=1 
    return even,odd

lst = [1,2,3,4,8,9,10]
even, odd = count(lst)         

print(even)
print(odd)
```
